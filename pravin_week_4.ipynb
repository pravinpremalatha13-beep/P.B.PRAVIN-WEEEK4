```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df =df.sort_values("date")
```

```python
df =df.set_index("date")
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=df,
    x=df.index,
    y="close"
)

plt.title("Shopify Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.show()
```

```python
#Visualize OHLC Price

plt.figure(figsize=(14,6))

sns.lineplot(data=df, x=df.index, y="open", label="open")

sns.lineplot(data=df, x=df.index, y="high", label="high")

sns.lineplot(data=df, x=df.index, y="low", label="low")

sns.lineplot(data=df, x=df.index, y="close", label="close")

plt.title("Shopify OHLC Prices Over Time")

plt.xlabel("Date")

plt.ylabel("Price")

plt.show()
```

```python
#Visualizing trading volume over time

plt.figure(figsize=(14,5))

sns.lineplot(

data=df,

x="date",

y="volume"
)

plt.title("Shopify Trading Volume Over Time")

plt.xlabel("Date")

plt.ylabel("Trading Volume")

plt.show()
```

```python
df.nlargest(5,"volume")
```

```python
#daily price change
df["daily price change"] = df["close"] - df["open"]
```

```python
#visualizing volume vs dailyplt.figure(figsize=(10,6))


sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)

plt.title("Trading volume vs daily price change")
plt.xlabel("volume")
plt.ylabel("daily price change")

plt.show()
```

```python
df["daily price range"] = df["close"]- df["open"]
df["daily price range"]
```

```python
#volume vs daily price range

plt.figure(figsize=(10,6))


sns.scatterplot(
    data=df,
    x="volume",
    y="daily price range"
)

plt.title("Trading volume vs daily price range")
plt.xlabel("volume")
plt.ylabel("daily price range")

plt.show()
```

```python
#Calculate the 20-Day SMA

df["SMA_20"] = df["close"].rolling(window=20).mean()
df["SMA_20"]
```

```python
#plot closing price + 20-Day SMA

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="Closing price"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_20",
    label="20-Day SMA"
)

plt.title("Shopify Closing Price and 20-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("SMA_20")

plt.show()
```

```python
#Calculate the 20-Day SMA

df["SMA_50"] = df["close"].rolling(window=50).mean()
df["SMA_50"]
```

```python
#plot closing price + 50-Day SMA

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="Closing price"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_50",
    label="50-Day SMA"
)

plt.title("Shopify Closing Price and 50-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("SMA_50")

plt.show()
```

```python
#Plot Close + Both Moving Averages

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="Closing price"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_20",
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_50",
    label="50-Day SMA"
)

plt.title("Shopify Closing Price with 20-Day and 50-day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#daily return%

df["daily return%"] = ((df["close"] - df["open"]) / df["open"]) * 100
```

```python
df["daily return%"].head()
```

```python

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="daily return%",
    bins=20,

)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")
```

```python


plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="daily return%",
    bins=20,
    kde=True #kernel density estimate
)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")

plt.show()
```

```python
#volatility

df["Absolute Return"] = df["daily return%"].abs()
df["Absolute Return"]
```

```python
df.nlargest(10,"Absolute Return")
```

```python
plt.figure(figsize=(14,5))

sns.lineplot(
data=df,
x="date",
y="daily return%"
)

plt.axhline(0,linestyle="--")
plt.xlabel("Date")
plt.ylabel("Daily Return%")
plt.title("Shopify daily returns over time")

plt.show()
```
